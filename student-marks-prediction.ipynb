# import the packages

```python
import numpy as np
import pandas as pd
import plotly.express as px
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("/kaggle/input/student-marks-dataset/Student_Marks.csv")
```

# finding the correlation

```python
correlation=data.corr()
print(correlation["Marks"].sort_values(ascending=False))
```

# seperating the target and features

```python
x = np.array(data[["time_study", "number_courses" ]])
y = np.array(data["Marks"])
```

# splitting the data into train and test sets

```python
x_train, x_test, y_train, y_test = train_test_split(x, y,  test_size=0.2, random_state=42)
```

# initialization of model

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
model.score(x_test, y_test)
```

```python
# Features = [["time_study", "number_courses"]]
features = np.array([[4.508, 8]])
model.predict(features)
```
